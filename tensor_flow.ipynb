```python
x=np.arange(-100,100,4)
y=np.arange(-90,110,4)
#making training test , testing set, validation set
x_train=x[:40]  #first 40 example of training the modle
y_train=y[:40]

x_test=x[40:]  #10 eamples for testing
y_test=y[40:]
```

```python
tf.random.set_seed(42)
model=tf.keras.Sequential([tf.keras.layers.Dense(1,input_shape=[1])])
model.compile(loss=tf.keras.losses.mae,
              optimizer=tf.keras.optimizers.SGD(),
              metrics=["mae"])
```

```python
model.summary()
```

```python
model.fit(x_train,y_train,epochs=200,verbose=0)
```

```python
y_predi=model.predict(x_test)
```

```python
y_predi
```

```python
def plot_predictions(train_data=x_train,
                     train_labels=y_train,
                     test_data=x_test,
                     test_labels=y_test,
                     predictions=y_predi):
  plt.figure(figsize=(10,7))
  plt.scatter(train_data,train_labels,c='r',label="training data")
  plt.scatter(test_data,test_labels,c='g',label="testing data")
  plt.scatter(test_data,predictions,c='b',label="predictions")
  plt.legend()
  plt.show()
```

```python
plot_predictions(train_data=x_train,
                     train_labels=y_train,
                     test_data=x_test,
                     test_labels=y_test,
                     predictions=y_predi)
```

```python
model.evaluate(x_test,y_test)
```

```python
tf.random.set_seed(42)
model_1=tf.keras.Sequential([tf.keras.layers.Dense(1)])
model_1.compile(loss=tf.keras.losses.mae,
                optimizer=tf.keras.optimizers.SGD(),
                metrics=['mae'])
model_1.fit(tf.expand_dims(x_train,axis=-1),y_train,epochs=100)
```

```python
y_preds_1=model_1.predict(x_test)
plot_predictions(predictions=y_preds_1)
```

```python
tf.random.set_seed(42)

# Replicate model_1 and add an extra layer
model_2 = tf.keras.Sequential([
  tf.keras.layers.Dense(1),
  tf.keras.layers.Dense(1) # add a second layer
])

# Compile the model
model_2.compile(loss=tf.keras.losses.mae,
                optimizer=tf.keras.optimizers.SGD(),
                metrics=['mae'])

# Fit the model
model_2.fit(tf.expand_dims(x_train, axis=-1), y_train, epochs=100, verbose=0)
```

```python
y_preds_2=model_2.predict(x_test)
plot_predictions(predictions=y_preds_2)
```

```python
tf.random.set_seed(42)

# Replicate model_2
model_3 = tf.keras.Sequential([
  tf.keras.layers.Dense(1),
  tf.keras.layers.Dense(1)
])

# Compile the model
model_3.compile(loss=tf.keras.losses.mae,
                optimizer=tf.keras.optimizers.SGD(),
                metrics=['mae'])

# Fit the model (this time for 500 epochs, not 100)
model_3.fit(tf.expand_dims(x_train, axis=-1), y_train, epochs=500, verbose=0) # set verbose to 0 for less output
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
insurance = pd.read_csv("https://raw.githubusercontent.com/stedy/Machine-Learning-with-R-datasets/master/insurance.csv")
```

```python
insurance.head()
```

```python
insurance_one_hot=pd.get_dummies(insurance)
insurance_one_hot.head()
```

```python
x=insurance_one_hot.drop("charges",axis=1)
y=insurance_one_hot["charges"]
```

```python
x.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
tf.random.set_seed(42)
insurance_model=tf.keras.Sequential([
    tf.keras.layers.Dense(1),
    tf.keras.layers.Dense(1)
])
insurance_model.compile(loss=tf.keras.losses.mae,
                        optimizer=tf.keras.optimizers.SGD(),
                        metrics=['mae']
                        )
insurance_model.fit(x_train,y_train,epochs=200)
```

```python
insurance_model.evaluate(x_test,y_test)
```

```python
#Our model didn't perform very well, let's try a bigger model.

#We'll try 3 things:

#Increasing the number of layers (2 -> 3).
#Increasing the number of units in each layer (except for the output layer).
#Changing the optimizer (from SGD to Adam).

tf.random.set_seed(42)
insurance_model_2=tf.keras.Sequential([
    tf.keras.layers.Dense(100),
    tf.keras.layers.Dense(10),
    tf.keras.layers.Dense(1)
])
insurance_model_2.compile(loss=tf.keras.losses.mae,
                        optimizer=tf.keras.optimizers.Adam(),
                        metrics=['mae']
                        )
history=insurance_model_2.fit(x_train,y_train,epochs=100,verbose=0)
```

```python
insurance_model_2.evaluate(x_test,y_test)
```

```python
pd.DataFrame(history.history).plot()
plt.ylabel("loss")
plt.xlabel("epochs");
```

```python
history_2=insurance_model_2.fit(x_train,y_train,epochs=100,verbose=0)
```

```python
insurance_model_2_loss,insurance_model_2_mae=insurance_model_2.evaluate(x_test,y_test)
insurance_model_2_loss,insurance_model_2_mae
```

```python
pd.DataFrame(history_2.history).plot()
plt.ylabel("loss")
plt.xlabel("epochs");
```

```python
from pathlib import Path
import pandas as pd
import tarfile
import urllib.request
def load_housing_data():
  tarball_path = Path("datasets/housing.tgz")
  if not tarball_path.is_file():
    Path("datasets").mkdir(parents=True, exist_ok=True)
    url = "https://github.com/ageron/data/raw/main/housing.tgz"
    urllib.request.urlretrieve(url, tarball_path)
    with tarfile.open(tarball_path) as housing_tarball:
      housing_tarball.extractall(path="datasets", filter='data')
  return pd.read_csv(Path("datasets/housing/housing.csv"))
housing = load_housing_data()
```
